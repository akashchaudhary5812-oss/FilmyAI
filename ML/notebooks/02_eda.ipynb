# 02 — Exploratory Data Analysis & Target Distributions
Analyzing commercial classifications, star power correlation, and temporal patterns.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ML.src.utils.config import get_project_root

root = get_project_root()
df = pd.read_parquet(root / 'data' / 'interim' / 'unified_movie_dataset.parquet')
print(df.info())
print(df['commercial_class_name'].value_counts())
```

```python
plt.figure(figsize=(8, 4))
df['commercial_class_name'].value_counts().plot(kind='bar', color='teal')
plt.title('Distribution of Commercial Success Categories')
plt.show()
```
